```python
import os.path

import commute_dm.utils
import momapy.io.core
import pd2af
```

```python
CD_AF_TRANSFORMATION_MODE = "keep-species"
CD_AF_LAYOUT_MODE = "plain"
CD_AF_INFLUENCE_PAIRING = "nearest"
```

```python
def cd_file_to_cd_af_file(
    input_file_path, output_file_path, mode, layout_mode, influence_pairing
):
    input_map = momapy.io.core.read(input_file_path).obj
    cd_af_map = pd2af.transform(
        input_map,
        mode=mode,
        layout_mode=layout_mode,
        influence_pairing=influence_pairing,
    ).obj
    momapy.io.core.write(cd_af_map, output_file_path, writer="celldesigner")


def make_cd_af_from_dir(input_dir, output_dir):
    for input_file_name, input_file_path in commute_dm.utils.list_dir(input_dir):
        print(input_file_name)
        output_file_name = commute_dm.utils.rename_file(input_file_name, "xml")
        output_file_path = os.path.join(output_dir, output_file_name)
        cd_file_to_cd_af_file(
            input_file_path,
            output_file_path,
            CD_AF_TRANSFORMATION_MODE,
            CD_AF_LAYOUT_MODE,
            CD_AF_INFLUENCE_PAIRING,
        )
```
