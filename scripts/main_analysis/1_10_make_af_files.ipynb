```python
import os.path

import commute_dm.utils
import momapy.io.core
import pd2af
```

```python
CD_AF_TRANSFORMATION_MODE = "keep-reactions"
# CD_AF_TRANSFORMATION_MODE = "normal"
CD_AF_LAYOUT_MODE = "plain"
# CD_AF_LAYOUT_MODE = "dot"
CD_AF_INFLUENCE_PAIRING = "nearest"
```

```python
def cd_file_to_cd_af_file(
    input_file_path, output_file_path, mode, layout_mode, influence_pairing
):
    # We read the source map together with its annotations (momapy keeps them in
    # a side-table on the ReaderResult, not on the map object), so pd2af can carry
    # them onto the transformed activity-flow elements and the writer can emit them.
    reader_result = momapy.io.core.read(input_file_path)
    transform_result = pd2af.transform(
        reader_result.obj,
        mode=mode,
        layout_mode=layout_mode,
        influence_pairing=influence_pairing,
        element_to_annotations=reader_result.element_to_annotations,
    )
    momapy.io.core.write(
        transform_result.obj,
        output_file_path,
        writer="celldesigner",
        element_to_annotations=transform_result.element_to_annotations,
    )


def make_cd_af_from_dir(input_dir, output_dir):
    for input_file_name, input_file_path in commute_dm.utils.list_dir(input_dir):
        print(input_file_name)
        output_file_name = commute_dm.utils.rename_file(input_file_name, "xml")
        output_file_path = os.path.join(output_dir, output_file_name)
        cd_file_to_cd_af_file(
            input_file_path,
            output_file_path,
            CD_AF_TRANSFORMATION_MODE,
            CD_AF_LAYOUT_MODE,
            CD_AF_INFLUENCE_PAIRING,
        )
```
