```python
import os.path

import cd2bel.core

import commute_dm.utils
```

```python
def make_bel_from_dir(input_dir, output_dir):
    for input_file_name, input_file_path in commute_dm.utils.list_dir(input_dir):
        print(input_file_name)
        output_file_name = commute_dm.utils.rename_file(input_file_name, "bel")
        bel_document_name = os.path.splitext(output_file_name)[0]
        bel_document_description = f"BEL file converted from {input_file_name} using the cd2bel package"
        output_file_path = os.path.join(output_dir, output_file_name)
        cd2bel.core.cd_file_to_bel_file(input_file_path, output_file_path, bel_document_name=bel_document_name, bel_document_description=bel_document_description)
```
